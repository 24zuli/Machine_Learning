# Encoder

```python
from sklearn.preprocessing import LabelEncoder
```

```python
label_encoders = {}
for column in df.select_dtypes(include='object').columns:
    label_encoders[column] = LabelEncoder()
    df[column] = label_encoders[column].fit_transform(df[column])
```

#### The target variable is Status, which tells whether a patient is Alive or Dead.

```python
X = df.drop(columns = 'Status')
```

```python
y = df['Status']
```

# Split Data

```python
from sklearn.model_selection import train_test_split,GridSearchCV, cross_val_score
from sklearn.metrics import classification_report
```

```python
X_train,X_test, y_train,y_test = train_test_split(X,y,test_size = 0.2, stratify=y,random_state = 42)
```

# Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(X_train, y_train)
dt_pred = dt_model.predict(X_test)
print("Decision Tree - Classification Report")
print(classification_report(y_test, dt_pred))
```

```python
dt_param_grid = {
    'max_depth': [3, 5, 10, None],
    'min_samples_split': [2, 5, 10]
}
dt_model = GridSearchCV(DecisionTreeClassifier(random_state=42), dt_param_grid, cv=5, scoring='accuracy', n_jobs=-1)
dt_model.fit(X_train, y_train)
print("Decision Tree Best Parameters:", dt_model.best_params_)
dt_pred = dt_model.predict(X_test)
print("Decision Tree - Classification Report")
print(classification_report(y_test, dt_pred))
```

#### A basic tree-based model.
#### It was optimized using Grid Search to find the best parameters like max_depth and min_samples_split.
#### Cross-validation ensured robust results.

# Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr_model = LogisticRegression(max_iter=1000, random_state=42)
lr_model.fit(X_train, y_train)
lr_pred = lr_model.predict(X_test)
print("Logistic Regression - Classification Report")
print(classification_report(y_test, lr_pred))
```

```python
lr_model = LogisticRegression(max_iter=1000, random_state=42)
cv_scores = cross_val_score(lr_model, X_train, y_train, cv=5, scoring='accuracy')
print(f"Logistic Regression Cross-Validation Accuracy: {cv_scores.mean():.4f}")
lr_model.fit(X_train, y_train)
lr_pred = lr_model.predict(X_test)
print("Logistic Regression - Classification Report")
print(classification_report(y_test, lr_pred))
```

#### A linear model used for binary classification.
#### Cross-validation was performed to measure accuracy across multiple splits of the data.


# RandomForest 

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf_model = RandomForestClassifier(random_state=42)
rf_model.fit(X_train, y_train)
rf_pred = rf_model.predict(X_test)
print("Random Forest - Classification Report")
print(classification_report(y_test, rf_pred))
```

```python
rf_param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [5, 10, None],
    'min_samples_split': [2, 5, 10]
}
rf_model = GridSearchCV(RandomForestClassifier(random_state=42), rf_param_grid, cv=5, scoring='accuracy', n_jobs=-1)
rf_model.fit(X_train, y_train)
print("Random Forest Best Parameters:", rf_model.best_params_)
rf_pred = rf_model.predict(X_test)
print("Random Forest - Classification Report")
print(classification_report(y_test, rf_pred))
```

#### An ensemble model made of multiple Decision Trees.
#### Grid Search was used to tune hyperparameters like n_estimators (number of trees) and max_depth.
#### Cross-validation improved the accuracy estimation.
